```python
import pandas as pd
from sqlalchemy import create_engine

# 1. 데이터 읽기 (일단 100개만 맛보기로 읽기)
url = "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2025-11.parquet"
print("데이터 다운로드 중...")
df = pd.read_parquet(url)

# 2. DB 연결
engine = create_engine("postgresql://root:root@localhost:5432/ny_taxi")

# 3. 딱 100개만 먼저 넣어서 테이블 생성 강제 확인!
print("딱 100개만 먼저 넣습니다...")
df.head(100).to_sql(name='yellow_tripdata_2025_11', con=engine, if_exists='replace', index=False)

print("✅ 이제 pgAdmin 새로고침(Refresh) 해보세요! 테이블이 생겼을 겁니다.")

# 4. 확인되었다면 나머지 다 넣기 (이미 생성된 테이블에 추가)
print("나머지 전체 데이터 적재 시작...")
df.iloc[100:].to_sql(name='yellow_tripdata_2025_11', con=engine, if_exists='append', index=False, chunksize=10000, method='multi')
```

```python
# 1마일 이하인 데이터 개수 바로 확인 (Question 3)
ans_q3 = len(df[
    (df['tpep_pickup_datetime'] >= '2025-11-01') & 
    (df['tpep_pickup_datetime'] < '2025-12-01') & 
    (df['trip_distance'] <= 1.0)
])

print(f"Question 3 정답: {ans_q3}")

# 가장 먼 거리를 간 날짜 확인 (Question 4)
longest_trip = df.loc[df['trip_distance'].idxmax()]
print(f"Question 4 날짜: {longest_trip['tpep_pickup_datetime'].date()}")
print(f"가장 먼 거리: {longest_trip['trip_distance']}")
```

```python
# 1. 먼저 2025년 11월 데이터만 별도의 데이터프레임으로 분리
df_nov_2025 = df[
    (df['tpep_pickup_datetime'] >= '2025-11-01') & 
    (df['tpep_pickup_datetime'] < '2025-12-01')
].copy()

# 2. 그 중에서 거리 조건(1마일 이하) 적용
ans_q3 = len(df_nov_2025[df_nov_2025['trip_distance'] <= 1.0])

print(f"2025년 11월 전체 데이터 수: {len(df_nov_2025)}")
print(f"Question 3 정답 (1마일 이하): {ans_q3}")
```
