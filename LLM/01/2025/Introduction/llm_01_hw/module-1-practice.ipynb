```python
import requests  # 인터넷에서 파일을 가져오기 위한 라이브러리 로드

# 1. 다운로드할 파일의 원본 주소 (GitHub에 저장된 minsearch.py의 원본 경로)
url = 'https://raw.githubusercontent.com/alexeygrigorev/minsearch/main/minsearch.py'

# 2. 해당 URL에 접속하여 파일 데이터를 요청(get)하고 응답을 받음
response = requests.get(url)

# 3. 로컬 컴퓨터에 'minsearch.py'라는 이름의 파일을 생성
# 'wb'는 Write Binary의 약자로, 텍스트뿐만 아니라 데이터 형태 그대로 쓰겠다는 의미입니다.
with open('minsearch.py', 'wb') as f:
    f.write(response.content)  # 인터넷에서 받아온 실제 내용(content)을 파일에 기록

print("✅ minsearch.py Download complete!") # 작업이 끝났음을 알리는 메시지
```

```python
# 1. 검색 엔진 도구(minsearch.py)를 가져옵니다.
import minsearch
import json  # JSON 파일을 다루기 위한 파이썬 기본 도구를 불러옵니다.

# 2. 'documents.json' 파일을 컴퓨터 메모리로 읽어옵니다.
# 'rt'의 의미: 
#   - 'r' (Read): 읽기 전용으로 열겠다.
#   - 't' (Text): 이미지나 실행 파일이 아닌, 글자(텍스트) 파일로 읽겠다.
# f_in: 파일을 가리키는 '통로' 이름을 관습적으로 f_in(File Input)이라고 지은 것입니다.

with open('documents.json', 'rt') as f_in:
    # json.load(f_in): 통로(f_in)를 통해 들어온 텍스트를 파이썬이 이해하는 '리스트와 딕셔너리' 구조로 자동 변환합니다.
    docs_raw = json.load(f_in)

# 3. 모든 데이터를 한 줄씩 평평하게 펼쳐서 담을 빈 주머니를 만듭니다.
documents = []

# 4. 큰 상자(코스 단위)를 하나씩 꺼냅니다. (예: 데이터 엔지니어링 코스 상자)
for course_dict in docs_raw:
    # course_dict['course']에는 "data-engineering-zoomcamp" 같은 코스 이름이 들어있습니다.
    course_name = course_dict['course']
    
    # 5. 큰 상자 안에 담긴 작은 봉투(개별 질문/답변 문서)들을 하나씩 꺼냅니다.
    for doc in course_dict['documents']:
        # 6. [중요] 작은 봉투(doc)에는 자기가 어느 코스 출신인지 적혀있지 않습니다.
        # 그래서 부모 상자에서 확인한 코스 이름(course_name)을 봉투 겉면에 'course'라는 이름표로 붙여줍니다.
        doc['course'] = course_name
        
        # 7. 이제 이름표가 붙은 봉투를 전체 목록(documents)에 차곡차곡 쌓습니다.
        documents.append(doc)

# 8. 작업이 잘 끝났는지 숫자로 확인합니다.
print(f"정제 완료! 전체 문서 개수: {len(documents)}개")
print("-" * 30)

# 9. 첫 번째 문서가 어떻게 바뀌었는지(이름표가 잘 붙었는지) 샘플을 확인합니다.
print("첫 번째 문서 샘플 (이름표 확인용):")
print(documents[0])
```

```python
# 1. minsearch 엔진 안에 있는 'Index'라는 기계를 꺼내서 설정을 시작합니다.
# 이 기계는 나중에 사용자의 질문을 받으면 어디를 뒤져야 할지 미리 약속하는 단계입니다.
index = minsearch.Index(
    # text_fields: "내용 기반 검색"을 할 구역입니다.
    # 사용자가 던진 질문의 '의미'를 파악해서 이 세 곳(질문, 본문, 섹션)에서 유사한 단어를 찾습니다.
    text_fields=["question", "text", "section"],
    
    # keyword_fields: "정확한 필터링"을 할 구역입니다.
    # 의미를 따지지 않고, SQL의 'WHERE' 절처럼 정확히 일치하는 데이터만 골라낼 때 사용합니다.
    # 여기서는 "어느 코스인지"만 딱 구분하기 위해 'course'를 지정했습니다.
    keyword_fields=["course"]
)

# 2. .fit(documents): 위에서 정한 규칙대로 실제 데이터(948개의 문서)를 공부시킵니다.
# 'fit'은 "딱 맞추다"라는 뜻으로, 엔진이 데이터를 훑으며 단어들의 점수판(TF-IDF 행렬)을 
# 완성하여 메모리에 올리는 과정입니다. 이 작업이 끝나야 비로소 검색할 준비가 끝납니다.
index.fit(documents)
```

```python
# 1. 사용자가 궁금해하는 질문(Query)을 문장으로 정의합니다.
query = 'the course has already started, can I still enroll?'

# 2. 부스팅(Boost) 설정: "어느 구역을 더 집중해서 볼까?"를 정합니다. 
# 단어가 발견되었을 때 '검색 점수'를 얼마나 뻥튀기할지 결정합니다.
# 질문(question) 필드에 3.0을 주면, 본문(text)보다 3배 더 중요하게 취급합니다.
# 섹션(section)에 0.5를 주면, 중요도를 절반으로 낮춰서 참고만 하겠다는 뜻입니다. 
# (기본값은 모두 1.0입니다.)
boost = {
    # 사용자의 질문 단어가 문서의 '질문' 칸에서 나오면 점수를 3.0배로 높게 쳐줍니다. (매우 중요)
    'question': 3.0, 
    
    # 사용자의 질문 단어가 문서의 '섹션' 칸에서 나오면 점수를 0.5배로 깎아서 계산합니다. (덜 중요)
    # 즉, 섹션 이름이 비슷하다고 해서 검색 순위가 확 올라가는 것을 방지합니다.
    'section': 0.5
}

# 3. 실제 검색 엔진 가동! (search 함수 호출)
results = index.search(
    query=query,                      # 사용자의 질문 내용을 전달합니다.
    filter_dict={'course': 'data-engineering-zoomcamp'}, # [필터] 지정한 코스 데이터만 뒤집니다.
    boost_dict=boost,                 # [가중치] 위에서 설정한 중요도를 적용합니다.
    num_results=5                     # [개수] 가장 점수가 높은 결과 5개만 가져옵니다.
)

# 4. 결과 확인: 엔진이 찾아온 5개의 문서 뭉치를 하나씩 꺼내서 화면에 출력합니다.
for res in results:
    # 각 결과(res)는 딕셔너리 형태이며, 질문/답변/코스 정보가 들어있습니다.
    print(res)
```

```python
def search(query):
    # 검색 시 제목(question)에 3배, 섹션에 0.5배의 가중치를 주는 설정입니다.
    boost = {'question': 3.0, 'section': 0.5}

    # 1.3장에서 인덱싱한 index 객체를 사용하여 검색을 수행합니다.
    results = index.search(
        query=query, # 사용자가 입력한 질문이 들어옵니다.
        filter_dict={'course': 'data-engineering-zoomcamp'}, # 특정 코스 데이터만 필터링합니다.
        boost_dict=boost, # 위에서 설정한 가중치를 적용합니다.
        num_results=5     # 상위 5개의 결과만 가져옵니다.
    )
    
    return results # 검색된 결과(문서 리스트)를 반환합니다.
```

```python
def build_prompt(query, search_results):
    # AI에게 줄 역할과 제약 사항이 담긴 설계도입니다.
    prompt_template = """
You're a course teaching assistant. Answer the QUESTION based on the CONTEXT from the FAQ database.
Use only the facts from the CONTEXT when answering the QUESTION.

QUESTION: {question}

CONTEXT: 
{context}
""".strip()

    # 검색된 문서들을 하나의 긴 문자열(Context)로 조립합니다.
    context = ""
    for doc in search_results:
        # 각 문서의 섹션, 질문, 답변을 구분해서 텍스트로 합칩니다. 
        # index -> results -> search_results -> doc 이며, index에는 section, question, text항목이 존재.
        context = context + f"section: {doc['section']}\nquestion: {doc['question']}\nanswer: {doc['text']}\n\n"
    
    # 설계도의 {question}과 {context} 자리에 실제 데이터를 집어넣습니다.
    prompt = prompt_template.format(question=query, context=context)
    return prompt # 완성된 최종 명령서를 반환합니다.
```

```python
# 제4장: LLM 답변 생성(LLM) 함수 정의 (4:02 ~ 4:33)
# "이제 완성된 프롬프트를 실제로 GPT-4o에게 전달하고 답변만 받아오는 llm 함수를 정의합니다."

# 💡 영상 4:08 ~ 4:24 구간 내용

def llm(prompt):
    # zakard님의 커스텀 클라이언트(get_openai_client)를 통해 요청을 보냅니다.
    response = client.chat.completions.create(
        model='gpt-4o',
        messages=[{"role": "user", "content": prompt}]
    )
    
    # 복잡한 응답 데이터 중에서 AI의 '답변 내용(content)'만 쏙 골라서 돌려줍니다.
    return response.choices[0].message.content
```

```python
query = 'how do I run kafka?'

def rag(query):
    search_results = search(query)
    prompt = build_prompt(query, search_results)
    answer = llm(prompt)
    return answer
```

```python
# [09:56] 테스트용 질문 정의
query = 'I just discovered the course. Can I still join it?'
```

```python
def elastic_search(query):
    
    # [10:20] 알렉세이가 설명하는 Elasticsearch 전용 검색 쿼리 객체 구조
    search_query = {
        "size": 5,  # 검색 결과 중 상위 5개만 가져오겠다는 설정 (SQL의 LIMIT 5)
        "query": {
            "bool": {  # 여러 조건을 조합할 때 사용하는 논리 블록 (AND, OR, FILTER 등)
                
                # 1. 'must' 블록: 검색 점수(Score)에 영향을 주는 '검색' 조건 (유사도 측정)
                "must": {
                    "multi_match": {
                        "query": query,  # 우리가 던진 질문(입력값)
                        
                        # [10:57] 여러 필드에서 검색하되, '^3'은 question 필드에서 단어가 발견되면
                        # 점수를 3배로 계산하라는 '가중치(Boosting)' 설정입니다.
                        "fields": ["question^3", "text", "section"], # [10:57] ^3은 3배 가중치를 의미함
                        
                        # 'best_fields': 지정된 필드 중 점수가 가장 높은 필드의 점수를 채택
                        "type": "best_fields"
                    }
                },
                
                # 2. 'filter' 블록: 점수 계산 없이 단순히 '맞다/아니다'로만 거르는 조건
                # 검색 속도가 매우 빠르고 캐싱이 되므로, 정확한 분류를 할 때 필수입니다.
                "filter": {
                    "term": {
                        # 인덱스 설정 시 'keyword' 타입으로 선언했기 때문에 정확히 일치해야 함
                        "course": "data-engineering-zoomcamp" # [10:40] 필터링 부분
                    }
                }
            }
        }
    }
    
    response = es_client.search(index=index_name, body=search_query)

    result_docs = []

    for hit in response['hits']['hits']:
        result_docs.append(hit['_source'])
    
    return result_docs
```

```python
def rag(query):
    search_results = elastic_search(query)
    prompt = build_prompt(query, search_results)
    answer = llm(prompt)
    return answer
```
