```python
#Q1

# File name: vector_search_homework.ipynb

import numpy as np
from fastembed import TextEmbedding

# 1. Initialize the embedding model by specifying the Jina model to use.
# (The first run will download and load the model, which may take a few seconds.)
embedding_model = TextEmbedding(model_name="jinaai/jina-embeddings-v2-small-en")

# 2. The query sentence provided in the homework.
query = 'I just discovered the course. Can I join now?'

# 3. Convert the text into a vector.
# Since fastembed's .embed() returns a Generator, 
# wrap it with list() to extract the actual values and fetch the first result ([0]).
query_vector = list(embedding_model.embed([query]))[0]

# 4. Print the dimension size (Shape) of the converted vector and the minimum value among all elements.
print(f"✅ Vector Shape: {query_vector.shape}")
print(f"🔥 Minimal Value: {query_vector.min():.4f}")
```

```python
# Cosine similarity
# File name: vector_search_homework.ipynb

# 1. Check the length (norm) of the query vector
# A norm of 1.0 means the vector is already normalized.
vector_norm = np.linalg.norm(query_vector)
print(f"✅ Vector Norm (Length): {vector_norm:.4f}")

# 2. Compute the dot product of the query vector with itself
# For normalized vectors, the dot product equals the cosine similarity.
self_dot_product = query_vector.dot(query_vector)
print(f"🔥 Dot Product with itself: {self_dot_product:.4f}")
```

```python
#Q2 
# File name: vector_search_homework.ipynb

# 1. Define the new document sentence provided in the homework.
doc = 'Can I still join the course after the start date?'

# 2. Convert the document text into a vector using the same Jina model.
doc_vector = list(embedding_model.embed([doc]))[0]

# 3. Compute the cosine similarity using the dot product.
# Since both vectors are already normalized, their dot product equals the cosine similarity.
cosine_sim = query_vector.dot(doc_vector)

# 4. Print the final cosine similarity value.
print(f"🔥 Cosine Similarity: {cosine_sim:.4f}")
```

```python
# File name: vector_search_homework.ipynb

# requests: Imports the HTTP library to download json data from the web.
import requests

# QdrantClient: Imports the main client to communicate with the Qdrant Vector DB.
from qdrant_client import QdrantClient

# Distance, VectorParams, PointStruct: Imports vector configuration parameters and data structures for Qdrant.
from qdrant_client.models import Distance, VectorParams, PointStruct

# TextEmbedding: Imports the core class from fastembed to convert text into high-dimensional vectors.
from fastembed import TextEmbedding

# 1. Download and filter the dataset (Preserved)
# docs_url: Remote GitHub repository raw URL containing the course document JSON data.
docs_url = 'https://github.com/alexeygrigorev/llm-rag-workshop/raw/main/notebooks/documents.json'

# requests.get(): Sends a GET request to download the raw data from the internet.
docs_response = requests.get(docs_url)

# .json(): Parses the response body into native Python lists and dictionaries.
documents_raw = docs_response.json()

# documents: Initializes an empty list to gather filtered target FAQ documents.
documents = []

# for course in documents_raw: Iterates through the entire dataset by top-level courses.
for course in documents_raw:
    # course_name: Extracts the name of the current course in the loop.
    course_name = course['course']
    
    # if course_name != ...: Filters out all other course logs that do not belong to the target course.
    if course_name != 'machine-learning-zoomcamp':
        # continue: Skips the rest of the loop block and moves to the next course if it's a mismatch.
        continue
        
    # for doc in ...: Loops through each individual FAQ document inside the filtered machine-learning-zoomcamp.
    for doc in course['documents']:
        # doc['course'] = ...: Explicitly attaches the course metadata field to the individual document dictionary.
        doc['course'] = course_name
        
        # documents.append(doc): Appends the fully processed document into the final collection list.
        documents.append(doc)

# 2. Load the BAAI/bge-small-en model from Q5 (384 dimensions) (Preserved)
# embedding_model: Loads the lightweight BAAI embedding model (384-dim) verified in Q5 into system memory.
embedding_model = TextEmbedding(model_name="BAAI/bge-small-en")

# 3. Create an in-memory Qdrant client and initialize the collection (Preserved)
# client: Generates a volatile, ultra-fast temporary Qdrant client running strictly inside RAM.
client = QdrantClient(":memory:")

# collection_name: Defines the bucket/table name to group related data points inside Qdrant.
collection_name = "ml_zoomcamp_faq"

# client.create_collection(): Initializes a new operational collection block inside the local Qdrant instance.
client.create_collection(
    # collection_name: Sets the identifier for this new vector table space to "ml_zoomcamp_faq".
    collection_name=collection_name,
    # vectors_config: Locks down the spatial constraints (size=384) and similarity metric (COSINE) for storing vectors.
    vectors_config=VectorParams(size=384, distance=Distance.COSINE),
)

# 4. Embed data and convert into Qdrant PointStruct format (Preserved)
# points: A container list to store properly structured PointStruct records prior to a batch upsert.
points = []

# for idx, doc in ...: Loops through the documents while concurrently unpacking indices (idx) and objects (doc).
for idx, doc in enumerate(documents):
    # Combine question and text fields as instructed (Preserved)
    # full_text: Concatenates the question and text strings with an intervening space character as per instruction.
    full_text = doc['question'] + ' ' + doc['text']
    
    # Extract the text embedding vector (Preserved)
    # vector: Evaluates the generator stream via list() and extracts the primary 1D numerical array at index 0.
    vector = list(embedding_model.embed([full_text]))[0]
    
    # points.append(): Converts the standard dictionary into a formal PointStruct record and adds it to the list.
    points.append(
        PointStruct(
            # id: Assigns a unique integer index acting as the primary key for the vector entry.
            id=idx,
            # vector: Binds the computed 384-dimensional dense floating-point matrix coordinates.
            vector=vector,
            # payload: Attaches the original source text fields (question, answer text, etc.) next to the vector.
            payload=doc
        )
    )

# 5. Upsert points into Qdrant collection (Preserved)
# client.upsert(): Commits the formatted list of vector points into the targeted Qdrant storage table at once.
client.upsert(collection_name=collection_name, points=points)

# 6. Define the query from Q1 and execute search (Preserved)
# query: Declares the identical search phrase string submitted back in Q1.
query = "I just discovered the course. Can I still join it?"

# query_vector: Converts the raw query text into a matching 384-dimensional mathematical vector to enable spatial calculation.
query_vector = list(embedding_model.embed([query]))[0]

# === Q6 FIX: qdrant-client 1.14+ has no client.search() — use query_points ===
search_response = client.query_points(
    collection_name=collection_name,
    query=query_vector,
    limit=1,
)
search_results = search_response.points

# 7. Print the highest similarity score (Preserved)
highest_score = search_results[0].score

# print(): Outlines the final maximum cosine similarity score generated against the database.
print(f"🔥 Highest similarity score for the first record: {highest_score}")
```
