## Creating arrays


```python
# 길이 10인 배열을 만들고, 모든 값을 0.0 으로 채웁니다.
np.zeros(10)
```

```python
# 길이 10인 배열을 만들고, 모든 값을 1.0 으로 채웁니다.
np.ones(10)
```

```python
# 길이 10인 배열을 만들고, 모든 값을 2.5 로 채웁니다.
# 첫 인자: 길이(크기), 둘째 인자: 채울 값
np.full(10, 2.5)
```

```python
# 파이썬 리스트를 NumPy 배열로 변환합니다.
a = np.array([1, 2, 3, 5, 7, 12])
# 변수 a 를 출력해 배열 내용을 확인합니다.
a
```

```python
# 인덱스 2 위치(세 번째 값, 원래 3)를 10 으로 바꿉니다.
# 인덱스는 0부터 시작합니다: a[0]=1, a[1]=2, a[2]=3 ...
a[2] = 10
```

```python
# 값이 바뀌었는지 배열 전체를 다시 확인합니다.
a
```

```python
# 3 이상 10 미만의 정수를 순서대로 만든 배열입니다. (3,4,5,6,7,8,9)
# 끝값 10 은 포함되지 않습니다. (파이썬 range 와 같음)
np.arange(3, 10)
```

```python
# 0 부터 100 까지를 같은 간격으로 나눈 숫자 11개를 만듭니다.
# (시작, 끝, 개수) → 0, 10, 20, ... , 100
np.linspace(0, 100, 11)
```

## Multi-dimensional arrays


```python
# 5행 2열(세로 5, 가로 2) 크기의 2차원 배열을 만들고 전부 0 으로 채웁니다.
# 인자를 튜플 (행, 열) 로 줍니다.
np.zeros((5, 2))
```

```python
# 3행 3열 행렬을 직접 만듭니다. 바깥 [] 가 행(row), 안쪽 [] 가 한 행의 값들입니다.
n = np.array([
    [1, 2, 3],  # 0번 행
    [4, 5, 6],  # 1번 행
    [7, 8, 9]   # 2번 행
])
```

```python
# n[행, 열] 형식입니다. 0번 행, 1번 열의 값(원래 2)을 20 으로 바꿉니다.
n[0, 1] = 20
```

```python
# 행렬 n 전체를 출력해 변경 결과를 확인합니다.
n
```

```python
# 2번 행 전체를 [1, 1, 1] 로 바꿉니다. (한 행을 통째로 교체)
n[2] = [1, 1, 1]
```

```python
# 2번 행이 바뀌었는지 확인합니다.
n
```

```python
# : 는 "모든 행" 이라는 뜻입니다. 모든 행의 2번 열을 [0, 1, 2] 로 바꿉니다.
# 즉 열 하나를 통째로 교체합니다.
n[:, 2] = [0, 1, 2]
```

```python
# 마지막 열이 바뀌었는지 확인합니다.
n
```

## Randomly generated arrays


```python
# seed(2): 난수 시작값을 고정합니다. 같은 seed 면 실행할 때마다 같은 난수가 나옵니다.
np.random.seed(2)
# 0~1 사이 난수로 5행 2열 배열을 만든 뒤, 100을 곱해 대략 0~100 범위로 키웁니다.
100 * np.random.rand(5, 2)
```

```python
# 난수 시드를 다시 2로 맞춥니다.
np.random.seed(2)
# 표준정규분포(평균 0, 분산 1)에서 뽑은 난수로 5행 2열 배열을 만듭니다.
# rand 와 달리 음수도 나올 수 있습니다.
np.random.randn(5, 2)
```

```python
# 난수 시드를 다시 2로 맞춥니다.
np.random.seed(2)
# 정수 난수를 만듭니다.
# low=0: 최솟값(포함), high=100: 최댓값(미포함), size=(5,2): 5행 2열
np.random.randint(low=0, high=100, size=(5, 2))
```

## Element-wise operations


```python
# 0 이상 5 미만 정수 배열 [0, 1, 2, 3, 4] 를 만듭니다.
a = np.arange(5)
# 배열 a 를 확인합니다.
a
```

```python
# 원소마다(element-wise) 같은 계산을 한 번에 적용합니다.
# 1) a * 2  → 각 원소에 2를 곱함
# 2) 10 + ... → 각 원소에 10을 더함
# 3) (... ) ** 2 → 각 원소를 제곱
# 4) / 100 → 각 원소를 100으로 나눔
b = (10 + (a * 2)) ** 2 / 100
```

```python
# 계산 결과 배열 b 를 확인합니다.
b
```

```python
# a 의 각 원소를 b 의 같은 위치 원소로 나눈 뒤, 전체에 10을 더합니다.
# (배열끼리 위치별로 연산)
a / b + 10
```

## Comparison operations

```python
# 비교에 쓸 배열 a 를 다시 확인합니다. [0, 1, 2, 3, 4]
a
```

```python
# a 의 각 원소가 2 이상인지 True/False 로 알려줍니다.
# 결과 예: [False, False, True, True, True]
a >= 2
```

```python
# 비교 대상 배열 b 를 확인합니다.
b
```

```python
# a 와 b 를 같은 위치끼리 비교합니다. a 원소가 b 원소보다 크면 True
a > b
```

```python
# 대괄호 안 조건이 True 인 위치의 a 값만 골라냅니다. (불리언 인덱싱 / 필터링)
# 즉 "b 보다 큰 a 의 원소들만" 남깁니다.
a[a > b]
```

## Summarizing operations

```python
# 요약 통계에 쓸 배열 a 를 확인합니다.
a
```

```python
# a 전체 원소의 표준편차(값이 평균에서 얼마나 퍼져 있는지)를 계산합니다.
a.std()
```

```python
# 행렬 n 에 들어 있는 모든 값 중 최솟값을 구합니다.
n.min()
```
