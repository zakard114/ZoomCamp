## DataFrames

```python
# 자동차 제원 정보를 2차원 리스트로 정의합니다. 각 안쪽 리스트 = 한 대의 차량(한 행).
data = [
    ['Nissan', 'Stanza', 1991, 138, 4, 'MANUAL', 'sedan', 2000],  # 1991년식 닛산 스탠자
    ['Hyundai', 'Sonata', 2017, None, 4, 'AUTOMATIC', 'Sedan', 27150],  # 2017 소나타 (Engine HP 누락 → None)
    ['Lotus', 'Elise', 2010, 218, 4, 'MANUAL', 'convertible', 54990],  # 2010 로터스 엘리스
    ['GMC', 'Acadia',  2017, 194, 4, 'AUTOMATIC', '4dr SUV', 34450],  # 2017 GMC 아카디아
    ['Nissan', 'Frontier', 2017, 261, 6, 'MANUAL', 'Pickup', 32340],  # 2017 닛산 프론티어
]

# 각 열(Column)에 붙일 이름을 순서대로 적은 문자열 리스트입니다.
columns = [
    'Make', 'Model', 'Year', 'Engine HP', 'Engine Cylinders',
    'Transmission Type', 'Vehicle_Style', 'MSRP'
]
```

```python
# 2차원 데이터와 컬럼 이름을 합쳐 판다스 데이터프레임(df)을 만듭니다.
df = pd.DataFrame(data, columns=columns)
```

```python
# 데이터프레임을 그대로 출력하면 표 형태로 정렬되어 보입니다.
df
```

```python
# 각 행을 '열 이름(키): 값' 딕셔너리로 두고, 그 딕셔너리들을 리스트로 묶은 형태입니다.
data = [
    {
        "Make": "Nissan",  # 제조사
        "Model": "Stanza",  # 모델명
        "Year": 1991,  # 연식
        "Engine HP": 138.0,  # 엔진 마력
        "Engine Cylinders": 4,  # 실린더 개수
        "Transmission Type": "MANUAL",  # 변속기
        "Vehicle_Style": "sedan",  # 차체 스타일
        "MSRP": 2000  # 권장소비자가격
    },
    {
        "Make": "Hyundai",
        "Model": "Sonata",
        "Year": 2017,
        "Engine HP": None,  # 누락 값은 파이썬 None 으로 표시
        "Engine Cylinders": 4,
        "Transmission Type": "AUTOMATIC",
        "Vehicle_Style": "Sedan",
        "MSRP": 27150
    },
    {
        "Make": "Lotus",
        "Model": "Elise",
        "Year": 2010,
        "Engine HP": 218.0,
        "Engine Cylinders": 4,
        "Transmission Type": "MANUAL",
        "Vehicle_Style": "convertible",
        "MSRP": 54990
    },
    {
        "Make": "GMC",
        "Model": "Acadia",
        "Year": 2017,
        "Engine HP": 194.0,
        "Engine Cylinders": 4,
        "Transmission Type": "AUTOMATIC",
        "Vehicle_Style": "4dr SUV",
        "MSRP": 34450
    },
    {
        "Make": "Nissan",
        "Model": "Frontier",
        "Year": 2017,
        "Engine HP": 261.0,
        "Engine Cylinders": 6,
        "Transmission Type": "MANUAL",
        "Vehicle_Style": "Pickup",
        "MSRP": 32340
    }
]
```

```python
# 딕셔너리 리스트로 데이터프레임을 만듭니다. 딕셔너리 키가 컬럼명이 됩니다.
df = pd.DataFrame(data)
# 생성된 데이터프레임을 출력합니다.
df
```

```python
# 위에서부터 n개 행만 미리 봅니다. (여기선 2행)
df.head(n=2)
```

## Series

```python
# 점(.) 표기법은 열 이름에 공백이 있으면 문법 오류가 납니다. (의도된 실패 예시)
df.Engine HP
```

```python
# 대괄호로 열 이름을 지정하면 공백이 있어도 됩니다. 결과는 1차원 시리즈(Series)입니다.
df['Engine HP']
```

## Index


```python
# 행을 구분하는 인덱스(기본: 0부터 시작하는 정수)를 조회합니다.
df.index
```

```python
# 'Make' 시리즈가 가진 인덱스도 조회합니다. (보통 데이터프레임 인덱스와 같습니다)
df.Make.index
```

```python
# 행 인덱스 라벨을 알파벳 'a'~'e'로 직접 지정합니다.
df.index = ['a', 'b', 'c', 'd', 'e']
```

```python
# 인덱스가 알파벳으로 바뀐 데이터프레임을 확인합니다.
df
```

```python
# iloc: 라벨 이름과 무관하게, 위치 기준(0부터)으로 1·2·4번째 행을 고릅니다.
df.iloc[[1, 2, 4]]
```

```python
# 문자열 인덱스를 버리고(drop=True) 0부터 시작하는 기본 정수 인덱스로 되돌립니다.
df = df.reset_index(drop=True)
```

```python
# 인덱스가 기본 정수로 초기화된 데이터프레임을 확인합니다.
df
```

## Element-wise operations

```python
# 'Engine HP' 열의 모든 값에 각각 2를 곱합니다. (요소별 연산, for문 불필요)
df['Engine HP'] * 2
```

```python
# 각 행의 Year가 2015 이상인지 True/False 시리즈로 반환합니다.
df['Year'] >= 2015
```

## Filtering

```python
# Make가 'Nissan'인 행만 남깁니다. (불리언 조건으로 필터링)
df[
    df['Make'] == 'Nissan'
]
```

```python
# 조건을 소괄호로 감싼 뒤 & (그리고)로 연결합니다. | 는 또는입니다.
df[
    (df['Make'] == 'Nissan') & (df['Year'] >= 2015)
]
```

## String operations

```python
# 파이썬 기본 문자열 replace 예시: 공백을 언더스코어(_)로 바꿉니다.
'machine learning zoomcamp'.replace(' ', '_')
```

```python
# .str 접근자로 열 전체 텍스트를 소문자로 바꿉니다.
df['Vehicle_Style'].str.lower()
```

```python
# 공백→언더스코어 치환 후 소문자 변환을 이어서(chaining) 적용하고, 원래 열에 저장합니다.
df['Vehicle_Style'] = df['Vehicle_Style'].str.replace(' ', '_').str.lower()
```

```python
# 문자열 정리가 반영된 데이터프레임을 확인합니다.
df
```

## Missing values


```python
# 결측값(Null/NaN)이면 True. 열별로 True 개수를 합쳐 결측 개수를 봅니다.
df.isnull().sum()
```

## Grouping


`
# (참고) SQL로 쓰면 아래와 비슷한 의미입니다. 실행용 파이썬 코드가 아닙니다.
SELECT 
    transmission_type,
    AVG(MSRP)
FROM
    cars
GROUP BY
    transmission_type
`


```python
# Transmission Type별로 묶은 뒤, 각 그룹의 MSRP 최댓값을 구합니다.
df.groupby('Transmission Type').MSRP.max()
```
