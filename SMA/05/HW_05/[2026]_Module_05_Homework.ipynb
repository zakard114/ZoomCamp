## Q1 — Korea mid/small-cap disclosures

Local-language information edge on KRX names with thin English coverage.

1. **Market and scope.** KRX mid/small caps that file mainly in Korean on DART. Horizon about 5–20 trading days after a material filing.
2. **Investment thesis.** Korean filings can move the local view before thin English coverage catches up. The story is uneven information, not “Korea grows.”
3. **Potential edge.** Reading Korean DART text directly. English screeners lag on many mid/small names. Liquidity can kill the edge, so I would filter by trading value.
4. **Data and sources.** Checked: Yahoo Finance `.KS` prices. To verify: DART open API timestamps/types, KRX liquidity filters, optional BOK/Statistics Korea industry prints.
5. **Research approach.** Event study of cumulative abnormal return vs KOSPI after chosen filing types. Later trade rule: long only if the filing category is constructive and volume clears a floor. Module 4 fees would still apply in a sim.
6. **Validation and risks.** Benchmark KOSPI and a random-filing placebo. Reject if returns die after costs, if only illiquid names work, or if English wires already match Korean filing times.


## Q2 — Memory-semiconductor relative value

Pairs / co-movement idea. Different mechanism from Q1.

1. **Market and scope.** Samsung Electronics vs SK hynix first (KRX). Optional robustness pair on US listings later. Horizon: days to a few weeks while the spread is stretched.
2. **Investment thesis.** Shared DRAM/NAND cycles pull the names together. Temporary relative gaps should mean-revert while the cycle link holds. Not an “AI forever” directional bet.
3. **Potential edge.** Headline AI tape treats the names as separate stories. A spread plus a shared cycle indicator can mark a dislocation the single-name narrative misses.
4. **Data and sources.** Checked: Yahoo Finance prices. To verify: public memory spot/contract series, Korea semiconductor exports, clean corporate-action handling for the ratio.
5. **Research approach.** Rolling residual of log prices (or ECM). Enter when residual z-score exceeds ±2 and the cycle indicator has not broken. Exit near z≈0 or on a time stop. Skip when rolling correlation collapses — that is the broken-relationship flag.
6. **Validation and risks.** Benchmark buy-and-hold either leg and a plain ratio z-score. Reject if half-life is too long, costs eat the reversion, or a regime shift permanently changes the beta. Borrow cost matters on the short leg.


## Q3 — US CPI surprise vs rate-sensitive equities

Macro event idea. Different mechanism again from Q1 and Q2.

1. **Market and scope.** Liquid US rate-sensitive proxies first: REIT, utility, and regional-bank ETFs. Horizon: CPI print day through about five sessions.
2. **Investment thesis.** Module 3 made me want CPI *surprise* (actual minus consensus), not only the CPI level. Hotter/colder-than-expected prints should move rate-sensitive baskets vs SPY over a short window through discount-rate jumps.
3. **Potential edge.** Slow monthly feature sets often carry lagged CPI levels. Surprise and same-morning nowcasts are easier to underweight. The edge is print timing, not “utilities always win.”
4. **Data and sources.** Checked in notes: Cleveland Fed inflation nowcast, FRED CPI/rates, Yahoo Finance ETF prices. To verify: free consensus or nowcast-vs-print surprise construction, exact release timestamps, stable ETF list.
5. **Research approach.** Event study. surprise = actual − consensus (or actual − prior nowcast). Bucket by sign/size; long or short the rate-sensitive basket vs SPY for N days. Module 4 fee assumptions. A signed surprise table is enough for the first test.
6. **Validation and risks.** Benchmark SPY and random-Wednesday trades. Reject if nothing survives costs, if a regime cut kills it, or if the open already prices the print. Watch FOMC overlap weeks and ETF spreads.
